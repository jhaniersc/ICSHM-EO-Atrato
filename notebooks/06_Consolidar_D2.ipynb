```python
from google.colab import drive
import pandas as pd

drive.mount('/content/drive')
carpeta = '/content/drive/MyDrive/ACDSA_2027/'

df1 = pd.read_csv(carpeta + 'Atrato_D2_NDTI_chunk1.csv')
df2 = pd.read_csv(carpeta + 'Atrato_D2_NDTI_chunk2.csv')
df3 = pd.read_csv(carpeta + 'Atrato_D2_NDTI_chunk3.csv')

print(f"Chunk 1: {len(df1)} filas")
print(f"Chunk 2: {len(df2)} filas")
print(f"Chunk 3: {len(df3)} filas")

df_completo = pd.concat([df1, df2, df3], ignore_index=True)
print(f"\nTotal consolidado: {len(df_completo)} filas")

duplicados = df_completo['HYBAS_ID'].duplicated().sum()
print(f"HYBAS_ID duplicados: {duplicados}")
assert duplicados == 0, "ERROR: HYBAS_ID repetidos"
assert len(df_completo) == 170, f"ERROR: total {len(df_completo)} != 170"

validos = df_completo['D2_NDTI'].notna().sum()
faltantes = df_completo['D2_NDTI'].isna().sum()
print(f"\nSubcuencas con D2_NDTI valido: {validos}")
print(f"Subcuencas con D2_NDTI faltante (nubes o sin agua): {faltantes}")

print(f"\n=== ESTADISTICAS D2_NDTI ===")
print(df_completo['D2_NDTI'].describe())

print(f"\n=== NDTI por chunk ===")
print(df_completo.groupby('chunk_id')['D2_NDTI'].describe())

ruta_salida = carpeta + 'Atrato_D2_NDTI_consolidado.csv'
df_completo.drop(columns=['chunk_id']).to_csv(ruta_salida, index=False)
print(f"\nGuardado: {ruta_salida}")
```
